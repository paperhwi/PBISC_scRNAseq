## Real-only / floor-only pockets (30-NN in UMAP)

```python
import pandas as pd, numpy as np
from sklearn.neighbors import NearestNeighbors
rows=[]
for coh in ["sepsis","covid"]:
    o=pd.read_pickle(f"myeloid/{coh}_myeloid_umap.pkl"); ref=(o.source=="ref").values
    idx=NearestNeighbors(n_neighbors=31).fit(o[["u1","u2"]].values).kneighbors(return_distance=False)
    rf=ref[idx].mean(1); iso=(~ref)&(rf<0.1); isor=ref&(rf>0.9)
    print(coh,"floor-only pocket",iso.sum(),f"({iso.mean()/(~ref).mean():.1%} of floor)","| real-only pocket",isor.sum(),f"({isor.sum()/ref.sum():.1%} of real)")
    print("  real-only by subset:",o[isor]["sub"].value_counts().head(5).to_dict())
    print("  real-only by patient:",o[isor]["patient"].value_counts().head(5).to_dict())
    print("  floor-only by subset:",o[iso]["sub"].value_counts().head(5).to_dict())
```
