Runs the statistics step (s08) and the panel script with the environment variables described in README.md. Outputs were removed (repository convention); the executed copy is kept in the local deliverables folder.

```
import os, subprocess, pandas as pd
HERE=os.path.abspath('.')
os.environ.setdefault('RUN','/path/to/run/')   # s02 outputs: RUN/out/sample_level_composition_metadata.csv
os.environ.setdefault('FIG','/path/to/fig/')   # s06/s07 outputs: FIG/data_v3/*.csv
os.environ.setdefault('GAIN',HERE+'/gain/')    # s08 writes GAIN/data/*.csv
```

```
os.makedirs(os.environ['GAIN'],exist_ok=True)
# run s08 from inside GAIN so that its data/ folder lands there
subprocess.run(['python',HERE+'/code/s08_gain_analysis.py'],cwd=os.environ['GAIN'],check=True)
G=os.environ['GAIN']+'data/'
display(pd.read_csv(G+'gain_severity_rho.csv').round(4)); display(pd.read_csv(G+'gain_death_auc.csv').round(4)); display(pd.read_csv(G+'gain_n80.csv').round(3))
```

```
os.makedirs('fig7_out',exist_ok=True)
subprocess.run(['python',HERE+'/figure/make_panels_v5.py'],cwd='fig7_out',check=True)
subprocess.run(['python',HERE+'/code/s09_shared_patient_check.py'],cwd='fig7_out',check=True)
display(pd.read_csv('fig7_out/data/shared_patient_check.csv').round(3))
```
